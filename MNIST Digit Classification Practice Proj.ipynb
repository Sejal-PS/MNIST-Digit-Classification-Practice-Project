##  Title:MNIST Digit Classification Practice Project

```python
import tensorflow as tf

# 1. Load Datasets: MNIST dataset contains 60,000 28x28 grayscale images of the 10 digits.
mnist = tf.keras.datasets.mnist
(x_train, y_train), (x_test, y_test) = mnist.load_data()
```

```python
# 2. Normalize Pixel values to range [0, 1] as Scaling improves training stability and performance.
x_train, x_test = x_train / 255.0, x_test / 255.0
```

```python
# 3. Flatten 28*28 images into 784-length vectors - Reshaping the 2D image data into a 1D format for the Dense layer.
x_train = x_train.reshape(-1, 28 * 28)
x_test = x_test.reshape(-1, 28 * 28)
```

```python
# 4. Build a Simple Neural Network
model = tf.keras.Sequential([
    # Hidden layer with 128 neurons and ReLU activation
    tf.keras.layers.Dense(128, activation='relu'), 
    # Output layer with 10 neurons (one for each digit) and Softmax for probability
    tf.keras.layers.Dense(10, activation='softmax') 
])
```

```python
# 5. Compile the model
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
```

```python
# 6. Train the model
# Training for 5 epochs (full passes through the dataset)
model.fit(x_train, y_train, epochs=5)
```

```python
# 7. Evaluate the model
test_loss, test_acc = model.evaluate(x_test, y_test)
print(f"\nTest accuracy: {test_acc:.4f}")
```

```python
# 8.Visualization
import matplotlib.pyplot as plt

plt.figure(figsize=(10,2))
for i in range(5):
    plt.subplot(1, 5, i+1)
    plt.imshow(x_train[i].reshape(28, 28), cmap='gray')
    plt.title(f"Label: {y_train[i]}")
    plt.axis('off')
plt.show()
```

```python
# Result
import numpy as np

predictions = model.predict(x_test[:5])
for i in range(5):
    print(f"Predicted: {np.argmax(predictions[i])}, Actual: {y_test[i]}")
```
